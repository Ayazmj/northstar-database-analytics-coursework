# 00 Project Setup and Data Audit

## Project Title
Integrated Database and Analytics Solution for NorthStar Urban Mobility and Logistics

## Purpose
This notebook performs the initial dataset audit for the NorthStar case study. It checks file availability, dataset structure, missing values, duplicate records, data types, and initial data quality issues before SQL, R, Python analytics, and MongoDB development are carried out.

## Technologies Used
- Google Colab
- Python
- Pandas
- NumPy
- GitHub


```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
import pandas as pd
import numpy as np
import os

data_path = "/content/drive/MyDrive/NorthStar_Database_Analytics/data "

files = {
    "customers": "customers.csv",
    "orders": "orders.csv",
    "deliveries": "deliveries.csv",
    "drivers": "drivers.csv",
    "vehicles": "vehicles.csv",
    "hubs": "hubs.csv",
    "complaints": "complaints.csv",
    "incidents": "incidents.csv",
    "app_events": "app_events.csv",
    "data_dictionary": "data_dictionary.csv"
}

datasets = {}

for name, filename in files.items():
    file_path = os.path.join(data_path, filename)

    if os.path.exists(file_path):
        datasets[name] = pd.read_csv(file_path)
        print(f"{name} loaded successfully: {datasets[name].shape}")
    else:
        print(f"{name} NOT FOUND: {file_path}")
```

```python
audit_summary = []

for name, df in datasets.items():
    audit_summary.append({
        "dataset": name,
        "rows": df.shape[0],
        "columns": df.shape[1],
        "duplicate_rows": df.duplicated().sum(),
        "missing_values_total": df.isnull().sum().sum()
    })

audit_df = pd.DataFrame(audit_summary)
audit_df
```

## Initial Dataset Audit Summary

The initial data audit confirms that all required NorthStar dataset files were successfully loaded into Google Colab. The audit summary records the number of rows, columns, duplicate rows, and missing values in each dataset. This step is important because it helps identify data quality issues before performing SQL analysis, R analytics, Python processing, and MongoDB database development.

```python
for name, df in datasets.items():
    print("\n" + "="*60)
    print(f"DATASET: {name.upper()}")
    print("="*60)
    print(df.info())
```

```python
for name, df in datasets.items():
    print("\n" + "="*60)
    print(f"MISSING VALUES IN {name.upper()}")
    print("="*60)

    missing = df.isnull().sum()
    missing = missing[missing > 0].sort_values(ascending=False)

    if missing.empty:
        print("No missing values found.")
    else:
        print(missing)
```

## Initial Dataset Audit Findings

The initial dataset audit shows that all required NorthStar datasets were successfully loaded into Google Colab. No duplicate rows were found across the datasets, which suggests that record-level duplication is not a major issue at this stage.

However, missing values are present in several important operational datasets. The highest number of missing values appears in the app_events dataset, followed by customers, deliveries, orders, incidents, complaints, drivers, and vehicles. This supports the case study problem that NorthStar's data environment is fragmented and inconsistent, especially where mobile platform events and operational records need to be combined.

The hubs and data_dictionary datasets contain no missing values, meaning they can be used as reliable reference datasets during analysis. Further cleaning will be required before performing SQL queries, R analytics, Python feature engineering, and MongoDB document modelling.

```python
missing_details = []

for name, df in datasets.items():
    for column in df.columns:
        missing_count = df[column].isnull().sum()
        if missing_count > 0:
            missing_details.append({
                "dataset": name,
                "column": column,
                "missing_count": missing_count,
                "missing_percentage": round((missing_count / len(df)) * 100, 2)
            })

missing_details_df = pd.DataFrame(missing_details)
missing_details_df.sort_values(by="missing_count", ascending=False)
```

```python
output_path = "/content/drive/MyDrive/NorthStar_Database_Analytics/outputs/"

os.makedirs(output_path, exist_ok=True)

audit_df.to_csv(output_path + "initial_dataset_audit_summary.csv", index=False)
missing_details_df.to_csv(output_path + "missing_values_by_column.csv", index=False)

print("Audit outputs saved successfully.")
```
