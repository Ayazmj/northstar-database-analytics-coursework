## SQL Analysis Questions

Five SQL queries were designed to investigate key operational and customer experience issues in the NorthStar dataset. The queries focus on service type reliability, hub-level delivery performance, customer complaint burden, driver performance risk, and app platform reliability.

The SQL queries use joins, grouping, aggregation, ordering, and calculated fields to generate decision-support insights.

```r
install.packages("sqldf")
install.packages("RSQLite")
install.packages("gsubfn")
install.packages("proto")

library(sqldf)
library(RSQLite)
library(gsubfn)
library(proto)
library(readr)
library(ggplot2)
library(dplyr)

print("Required packages loaded successfully.")
```

## Query 1: Service Type Failure Rate

This query identifies which service types have the highest delivery failure rate. The `orders` dataset is joined with the `deliveries` dataset using `order_id`. The query calculates total deliveries, failed deliveries, failure rate percentage, and average customer rating by service type.

```r
test_query <- sqldf("
SELECT
    COUNT(*) AS total_orders
FROM orders
")

test_query
```

```r
query1_service_failure <- sqldf("
SELECT
    o.service_type,
    COUNT(d.delivery_id) AS total_deliveries,
    SUM(d.delivery_status_failure_flag) AS failed_deliveries,
    ROUND(AVG(d.delivery_status_failure_flag) * 100, 2) AS failure_rate_percentage,
    ROUND(AVG(d.customer_rating_post_delivery), 2) AS avg_customer_rating
FROM orders o
JOIN deliveries d
    ON o.order_id = d.order_id
GROUP BY o.service_type
ORDER BY failure_rate_percentage DESC
")

query1_service_failure
```

### Interpretation

The result shows which service types experience the highest proportion of failed deliveries. A higher failure rate suggests that the service category may require operational review, improved routing, better driver allocation, or clearer customer communication. The average customer rating helps compare whether failure-prone service types also receive lower customer satisfaction scores.

## Query 2: Hub-Level Delivery Reliability

This query evaluates hub-level delivery performance by combining deliveries with hub information. It calculates delivery volume, average route distance, average manual route overrides, failure rate, and average customer rating by hub.

```r
query2_hub_reliability <- sqldf("
SELECT
    h.hub_name,
    h.zone,
    h.hub_type,
    COUNT(d.delivery_id) AS total_deliveries,
    ROUND(AVG(d.route_distance_km), 2) AS avg_route_distance_km,
    ROUND(AVG(d.manual_route_override_count), 2) AS avg_manual_overrides,
    ROUND(AVG(d.delivery_status_failure_flag) * 100, 2) AS failure_rate_percentage,
    ROUND(AVG(d.customer_rating_post_delivery), 2) AS avg_customer_rating
FROM deliveries d
JOIN hubs h
    ON d.hub_id = h.hub_id
GROUP BY h.hub_name, h.zone, h.hub_type
ORDER BY failure_rate_percentage DESC, avg_manual_overrides DESC
")

query2_hub_reliability
```

### Interpretation

This query highlights hubs with weaker delivery reliability. Hubs with high failure rates and high manual route override averages may indicate routing inefficiencies, capacity pressure, or operational control issues. These hubs should be prioritised for process review and resource planning.

## Query 3: Customer Segment Complaint Burden

This query analyses complaint burden by customer type. It joins customers and complaints to calculate total complaints, high-severity complaints, average resolution days, and total compensation amount for each customer segment.

```r
query3_customer_complaints <- sqldf("
SELECT
    c.customer_type,
    COUNT(cp.complaint_id) AS total_complaints,
    SUM(cp.severity_high_flag) AS high_severity_complaints,
    ROUND(AVG(cp.resolution_days), 2) AS avg_resolution_days,
    ROUND(SUM(cp.compensation_amount), 2) AS total_compensation_amount,
    ROUND(AVG(cp.compensation_amount), 2) AS avg_compensation_amount
FROM customers c
JOIN complaints cp
    ON c.customer_id = cp.customer_id
GROUP BY c.customer_type
ORDER BY total_complaints DESC, total_compensation_amount DESC
")

query3_customer_complaints
```

### Interpretation

The result identifies which customer segments create the highest complaint and compensation burden. Segments with high complaint counts and high compensation values may require targeted service improvement, better communication, or priority support handling.

## Query 4: Driver Performance Risk

This query investigates driver performance by employment type and experience group. It joins drivers with deliveries and calculates delivery volume, failure rate, average customer rating, average route distance, average manual route overrides, average training score, and average driver rating.

```r
query4_driver_risk <- sqldf("
SELECT
    dr.employment_type,
    CASE
        WHEN dr.years_experience < 2 THEN 'Low Experience'
        WHEN dr.years_experience BETWEEN 2 AND 5 THEN 'Medium Experience'
        ELSE 'High Experience'
    END AS experience_group,
    COUNT(d.delivery_id) AS total_deliveries,
    ROUND(AVG(d.delivery_status_failure_flag) * 100, 2) AS failure_rate_percentage,
    ROUND(AVG(d.customer_rating_post_delivery), 2) AS avg_customer_rating,
    ROUND(AVG(d.route_distance_km), 2) AS avg_route_distance_km,
    ROUND(AVG(d.manual_route_override_count), 2) AS avg_manual_overrides,
    ROUND(AVG(dr.training_score), 2) AS avg_training_score,
    ROUND(AVG(dr.driver_rating), 2) AS avg_driver_rating
FROM drivers dr
JOIN deliveries d
    ON dr.driver_id = d.driver_id
GROUP BY dr.employment_type, experience_group
ORDER BY failure_rate_percentage DESC, avg_manual_overrides DESC
")

query4_driver_risk
```

### Interpretation

This query shows whether delivery reliability differs by employment type and driver experience. Higher failure rates among certain driver groups may indicate the need for additional training, better shift allocation, or improved route support.

## Query 5: App Platform Reliability

This query analyses app platform reliability using app event records. It groups events by device type and event type to calculate event volume, average API latency, success rate, and issue rate.

```r
query5_app_reliability <- sqldf("
SELECT
    device_type,
    event_type,
    COUNT(event_id) AS total_events,
    ROUND(AVG(api_latency_ms), 2) AS avg_api_latency_ms,
    ROUND(AVG(success_flag) * 100, 2) AS success_rate_percentage,
    ROUND((1 - AVG(success_flag)) * 100, 2) AS issue_rate_percentage
FROM app_events
GROUP BY device_type, event_type
ORDER BY issue_rate_percentage DESC, avg_api_latency_ms DESC
")

query5_app_reliability
```

### Interpretation

The result identifies app event categories and device types with lower success rates or higher API latency. This is important because poor app performance can affect booking reliability, customer experience, and operational visibility.

## SQL Query Optimisation Discussion

Although the SQL queries in this notebook are executed on in-memory R data frames using `sqldf`, query optimisation principles are still important. The following optimisation strategies were considered:

1. **Join only required tables**  
   Each query joins only the datasets needed to answer the business question.

2. **Select only relevant columns**  
   Queries avoid selecting unnecessary columns, reducing memory usage and improving readability.

3. **Use indexed join keys in database systems**  
   In a production relational database, columns such as `order_id`, `customer_id`, `driver_id`, `vehicle_id`, and `hub_id` should be indexed because they are frequently used in joins.

4. **Aggregate after joining only required records**  
   Grouping is performed after the relevant records are joined, which supports clearer and more efficient analysis.

5. **Export summary outputs**  
   The SQL outputs are exported as CSV files so that the results can be reused in the final report without rerunning all queries.
